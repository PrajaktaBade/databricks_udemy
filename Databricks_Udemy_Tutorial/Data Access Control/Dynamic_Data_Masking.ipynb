### **Create the Mask Function**

```python
%sql
CREATE OR REPLACE FUNCTION databricks_prajakta.silver.dynamic_mask(p_email STRING)
RETURNS STRING
RETURN (
  CASE 
    WHEN is_account_group_member('developers') THEN p_email 
    ELSE '***' 
  END
)
```

### **Applying Mask**

```python
%sql
alter table databricks_prajakta.silver.customers_enr
alter column email set mask databricks_prajakta.silver.dynamic_mask
```
