### **Mapping Table**

```python
%sql
create or replace table databricks_prajakta.silver.rls_mapping
(
  user_id string,
  region string
)
```

### **Create RLS function**

```python
%sql
CREATE OR REPLACE FUNCTION databricks_prajakta.silver.rls_function(
  p_region STRING
)
RETURNS BOOLEAN
RETURN (
  EXISTS (
    SELECT 1
    FROM databricks_prajakta.silver.rls_mapping
    WHERE user_id = current_user()
      AND region = lower(p_region)
  )
)
```

### **Applying function on region Column**

```python
%sql 
alter table databricks_prajakta.silver.stores_enr
set row filter databricks_prajakta.silver.rls_function on (region)   
```
